- Bài 5.5: 


```python
import sqlite3

conn = sqlite3.connect('bai5_5.db')
cursor = conn.cursor()

cursor.execute("""
CREATE TABLE IF NOT EXISTS nhan_vien (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    ten TEXT NOT NULL,
    tuoi INTEGER,
    luong REAL
)
""")

cursor.execute("""
INSERT INTO nhan_vien (ten, tuoi, luong)
VALUES 
('Nguyen Van A', 30, 20000000),
('Tran Van B', 40, 30000000)
""")

conn.commit()

cursor.execute("SELECT * FROM nhan_vien")
rows = cursor.fetchall()

for row in rows:
    print(row)

conn.close()
```

- Bài 5.10: 

```python
import sqlite3

conn = sqlite3.connect('product.db')
cursor = conn.cursor()

cursor.execute("""
CREATE TABLE IF NOT EXISTS product (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    name TEXT NOT NULL,
    price REAL NOT NULL,
    amount INTEGER NOT NULL
)
""")

conn.commit()
conn.close()
```

```python
import sqlite3

conn = sqlite3.connect('product.db')
cursor = conn.cursor()

# Thêm sản phẩm
cursor.execute(
    "INSERT INTO product (name, price, amount) VALUES (?, ?, ?)",
    ("Ban phim", 300000, 10)
)

# Hiển thị
cursor.execute("SELECT * FROM product")
for row in cursor.fetchall():
    print(row)

conn.commit()
conn.close()
```

- Bài 5.11: 

```python
import sqlite3

conn = sqlite3.connect('ql_nhan_vien.db')
cursor = conn.cursor()

cursor.execute("""
CREATE TABLE IF NOT EXISTS PHONG (
    id_phong INTEGER PRIMARY KEY,
    ten_phong TEXT NOT NULL
)
""")

cursor.execute("""
CREATE TABLE IF NOT EXISTS NHAN_VIEN (
    id INTEGER PRIMARY KEY,
    ho_ten TEXT NOT NULL,
    tuoi INTEGER NOT NULL,
    dia_chi TEXT NOT NULL,
    luong REAL NOT NULL,
    id_phong INTEGER,
    FOREIGN KEY (id_phong) REFERENCES PHONG(id_phong)
)
""")

conn.commit()
conn.close()
```
