```python
class Stack:
    def __init__(self, capacity):
        # capacity: sức chứa tối đa của ngăn xếp
        self.capacity = capacity
        self.data = [0.0] * capacity  # mảng float
        self.top = -1  # vị trí đỉnh (ban đầu chưa có phần tử)

    def isEmpty(self):
        return self.top == -1

    def isFull(self):
        return self.top == self.capacity - 1

    def push(self, value):
        if self.isFull():
            print("Ngăn xếp đã đầy, không thể thêm phần tử!")
        else:
            self.top += 1
            self.data[self.top] = float(value)

    def pop(self):
        if self.isEmpty():
            print("Ngăn xếp trống, không thể lấy phần tử!")
            return None
        else:
            value = self.data[self.top]
            self.top -= 1
            return value

    def __del__(self):

        del self.data



if __name__ == "__main__":
    s = Stack(1)
    s.push(1.5)
    print("Lấy ra phần tử:", s.pop())
    print("Ngăn xếp trống?", s.isEmpty())
    print("Ngăn xếp đầy?", s.isFull())
```

```python
class Date:
    def __init__(self, day, month, year):
        self.day = day
        self.month = month
        self.year = year
    def next_day(self):
        if self.day == 30 and self.month < 12:
            self.day = 1
            self.month += 1
        elif self.month == 12 and self.day == 30:
            self.day = 1
            self.month = 1
            self.year += 1
        else:
            self.day += 1
    def display(self):
        return f'Ngay tiep theo la: Ngay: {self.day} Thang: {self.month} Nam: {self.year}'        

if __name__ == '__main__':
    day = int(input())
    month = int(input())
    year = int(input())
    d = Date(day, month, year)
    d.next_day()
    print(d.display())

```

```python
import math
#  Lớp cơ sở: Đa giác
class DaGiac:
    def __init__(self, so_canh):
        self.so_canh = so_canh  # số cạnh của đa giác

    def hien_thi_thong_tin(self):
        print(f"Đây là đa giác có {self.so_canh} cạnh.")

    def chu_vi(self):
        return None

    def dien_tich(self):
        return None


#  Lớp Tứ giác kế thừa từ Đa giác
class TuGiac(DaGiac):
    def __init__(self):
        super().__init__(4)  # Gọi hàm khởi tạo của lớp cha (Đa giác) → có 4 cạnh

    def hien_thi_thong_tin(self):
        print("Đây là tứ giác  một đa giác có 4 cạnh.")
        print("Tổng các góc trong bằng 360°.")


#  Lớp Hình bình hành kế thừa từ Tứ giác
class HinhBinhHanh(TuGiac):
    def __init__(self, a, b, h):
        super().__init__()   # Gọi hàm khởi tạo của lớp cha (Tứ giác)
        self.a = a  # cạnh đáy
        self.b = b  # cạnh bên
        self.h = h  # chiều cao tương ứng với cạnh đáy

    def chu_vi(self):
        return 2 * (self.a + self.b)

    def dien_tich(self):
        return self.a * self.h

    def hien_thi_thong_tin(self):
        print("Đây là hình bình hành:")
        print(f" - Cạnh đáy a = {self.a}")
        print(f" - Cạnh bên b = {self.b}")
        print(f" - Chiều cao h = {self.h}")
        print(f" - Chu vi: {self.chu_vi()}")
        print(f" - Diện tích: {self.dien_tich()}")
        print(" - Hai cặp cạnh đối song song và bằng nhau.")


#  Lớp Hình chữ nhật kế thừa từ Hình bình hành
class HinhChuNhat(HinhBinhHanh):
    def __init__(self, a, b):
        # Với hình chữ nhật, chiều cao = b vì có góc vuông
        super().__init__(a, b, h=b)

    def hien_thi_thong_tin(self):
        print("Đây là hình chữ nhật:")
        print(f" - Chiều dài a = {self.a}")
        print(f" - Chiều rộng b = {self.b}")
        print(f" - Chu vi: {self.chu_vi()}")
        print(f" - Diện tích: {self.dien_tich()}")
        print(" - Có 4 góc vuông, 2 đường chéo bằng nhau.")


#  Lớp Hình vuông kế thừa từ Hình chữ nhật
class HinhVuong(HinhChuNhat):
    def __init__(self, a):
        super().__init__(a, a)  # Hình vuông có 4 cạnh bằng nhau

    def chu_vi(self):
        return 4 * self.a

    def dien_tich(self):
        return self.a ** 2

    def hien_thi_thong_tin(self):
        print("Đây là hình vuông:")
        print(f" - Cạnh a = {self.a}")
        print(f" - Chu vi: {self.chu_vi()}")
        print(f" - Diện tích: {self.dien_tich()}")
        print(" - 4 cạnh bằng nhau, 4 góc vuông, 2 đường chéo vuông góc và bằng nhau.")


#  Phần kiểm tra
if __name__ == "__main__":
    print("=== MỐI QUAN HỆ KẾ THỪA GIỮA CÁC HÌNH ===\n")

    h1 = DaGiac(5)
    h1.hien_thi_thong_tin()
    print()

    h2 = TuGiac()
    h2.hien_thi_thong_tin()
    print()

    h3 = HinhBinhHanh(6, 4, 3)
    h3.hien_thi_thong_tin()
    print()

    h4 = HinhChuNhat(8, 5)
    h4.hien_thi_thong_tin()
    print()

    h5 = HinhVuong(5)
    h5.hien_thi_thong_tin()

```
