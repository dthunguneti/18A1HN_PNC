```python
import requests
import xml.dom.minidom as minidom
import csv

def main():
    url = 'http://feeds.bbci.co.uk/news/rss.xml'

    # Tải nội dung xml 
    response = requests.get(url)
    response.encoding = 'utf-8'
    xml_data = response.text

    # Phân tích xml bằng minidom
    doc = minidom.parseString(xml_data)

    # Lấy all các phần tử <item> mỗi item là một tin tức : 
    items = doc.getElementsByTagName('item')

    new_list = [] # List news

    for item in items:
        title = item.getElementsByTagName('title')[0].firstChild.data
        description = item.getElementsByTagName('description')[0].firstChild.data
        link = item.getElementsByTagName('link')[0].firstChild.data
        guid = item.getElementsByTagName('guid')[0].firstChild.data

        new_list.append({
            'Title': title,
            'Description' : description,
            'Link' : link, 
            'Guid' : guid
        })
    for new in new_list:
        print(f'Title : {new['Title']}')
        print(f'Description : {new['Description']}')
        print(f'Link : {new['Link']}')  
        print(f'Guid : {new['Guid']}\n')

    # Ghi đè vao file csv: 
    with open('rss_news.csv', 'w', newline='', encoding= 'utf-8') as file:
        writer = csv.DictWriter(file, fieldnames= ['Title', 'Description', 'Link', 'Guid' ])
        writer.writeheader()
        writer.writerows(new_list)
    print("Đã lưu vào file CSV")

if __name__ == '__main__':
    main()










```
