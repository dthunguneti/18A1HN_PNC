```python
import json

json_1 = '{"Name": "Đinh Tiến Hưng", "Age": "12/10/2006", "Skill": "Python, AI"}'

json_2 = '{\"Name\": \"Nguyễn Thành Hưng\", \"Age\": \"12-10-2006\", \"Skill\": \"C, SQL\"}'

# Chuyển chuỗi Json thành đối tượng python: 
json_object_1 = json.loads(json_1)
json_object_2 = json.loads(json_2)

print(json_object_1)

print(json_object_2)
```
