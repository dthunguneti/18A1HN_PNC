```python
import requests

response = requests.get('https://jsonplaceholder.typicode.com/posts')

if response.status_code == 200:
    json_data = response.json()
    tong_post = 0
    for post in json_data:
        tong_post += 1
        print('===============================')
        print(f'UserID: {post["userId"]}')
        print(f'ID: {post['id']}')
        print(f'Tieu de: {post['title']}')
        print(f'Noi dung: {post['body']}')
        print('===============================')
    print(f'Tong so bai dang: {tong_post}')
else:
    print('Yeu cau khong thanh cong. Ma trang thai: ',response.status_code)
```
