```python
import xml.dom.minidom

def main():
    company = xml.dom.minidom.parse('Bai2.3_sample.xml')

    # Lay cac the cua company: 
    staff_list = company.getElementsByTagName('staff')

    # In thong tin cua phan tu con dau tien: 
    first_staff = staff_list[0]
    print(first_staff.tagName)

    # Lay cac phan tu con trong staff 
    name = first_staff.getElementsByTagName('name')[0].firstChild.data
    salary = first_staff.getElementsByTagName('salary')[0].firstChild.data
    print(f'Ten: {name}')
    print(f'Salary: {salary}')

    # In tat cả các phần tử con :
    for staff in staff_list:
        id = staff.getAttribute('id')
        name = staff.getElementsByTagName('name')[0].firstChild.data.strip()
        salary = staff.getElementsByTagName('salary')[0].firstChild.data.strip()
        print(f'ID: {id}  Name: {name}   Salary: {salary}')


if __name__ == '__main__':
    main()
```
