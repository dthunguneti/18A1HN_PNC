```python
import json as j

# Mở, đọc file thong_ke_vn.json:
with open('thong_ke_nv.json', "r", encoding="utf-8") as f:
    infor = j.load(f)
    for key, value in infor.items():
        print(f"{key}: {value}")

print()

# Lấy thông tin chung: 
ten_cty = infor['Tên công ty']
address = infor['Địa chỉ']
ds_phong = infor['Thông Tin']

# Tổng số nhân viên trong công ty:
sum_nhan_vien_cong_ty = 0 
for nhan_su in ds_phong:
    sum_nhan_vien_cong_ty += nhan_su['Số lượng nhân viên']


# In ra màn hình : 
print(f'Tên công ty: {ten_cty}')
print(f"Địa chỉ: {address}")

print("------Thống kê nhân viên theo đơn vị------")

for i, phong in enumerate(ds_phong, start=1):
    print(f"{i}. Tên đơn vị: {phong["Tên đơn vị"]}")
    print(f"    - Số nhân viên: {phong["Số lượng nhân viên"]}")
    print(f"    - Tỷ lệ: {(phong["Số lượng nhân viên"]/sum_nhan_vien_cong_ty)*100} %")
    





```
