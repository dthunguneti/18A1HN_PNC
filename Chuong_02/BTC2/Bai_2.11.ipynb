```python
import json
from datetime import *


def giao_dich():
    ds_giao_dich = []
    while True:

        ma_giao_dich = input('Nhap ma giao dich: ') 

        loai_giao_dich = input("Nhap loai giao dich (Tien/ Vang): ")

        if loai_giao_dich.lower() == 'tien':
            so_tien = float(input('Nhap so tien giao dich (VND): '))
        if loai_giao_dich.lower() == 'vang':
            so_tien = float(input('Nhap so luong vang: '))

        ngay_giao_dich = input('Nhap ngay giao dich (dd-mm-yyyy): ')

        gd = {
            'ma_giao_dich': ma_giao_dich,
            'loai_giao_dich': loai_giao_dich,
            'so_tien': so_tien,
            'ngay_giao_dich': ngay_giao_dich 
        }

        ds_giao_dich.append(gd)

        giao_dich_tiep = input('Ban co muon tiep tuc giao dich (y = co| n = khong): ')
        if giao_dich_tiep == 'n' or giao_dich_tiep == 'N':
            break
    
    return ds_giao_dich

def ghi_file_json(ds):
    now = datetime.now()
    ten_file = now.strftime('%Y-%m-%d-%H-%M-%S.json')

    with open(ten_file, 'w', encoding='utf-8') as f:
        json.dump(ds,f, ensure_ascii=False, indent=4)
    print(f'Da luu giao dich thanh cong!!!')


def main():
    ds = giao_dich()

    choice = input('Ban co muon luu giao dich khong (1: Co | 0: Khong): ')
    if choice == '1':
        ghi_file_json(ds)
    else:
        print('Giao dich khong luu')

if __name__ == '__main__':
    main()
    
        


```
