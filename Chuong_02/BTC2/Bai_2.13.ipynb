```python
import requests as R
url = 'https://jsonplaceholder.typicode.com/comments?postId=1'

response = R.get(url)

if response.status_code == 200:
    data = response.json()
    for i in range(3):
        print(f"Bài {i+1}:")
        print(f"ID: {data[i]['id']}")
        print(f"Tên người dùng: {data[i]['name']}")
        print(f"Email: {data[i]['email']}")
        print(f"Nội dung: {data[i]['body']}\n")
else:
    print('Yeu cau khong thanh cong. Ma trang thai la: ',response.status_code)
```
