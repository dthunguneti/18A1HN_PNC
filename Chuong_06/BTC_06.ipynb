```python
import threading

def print_thread_name():
    print("Luồng đang chạy:", threading.current_thread().name)

threads = []

for i in range(5):
    t = threading.Thread(target=print_thread_name, name=f"Thread-{i+1}")
    threads.append(t)
    t.start()

for t in threads:
    t.join()

print("Kết thúc Main Thread")
```

- Bài 6.2: 

```python
import threading
import time

def task(name):
    for i in range(3):
        print(f"{name} đang chạy bước {i}")
        time.sleep(1)

t1 = threading.Thread(target=task, args=("Task 1",))
t2 = threading.Thread(target=task, args=("Task 2",))

t1.start()
t2.start()

t1.join()
t2.join()

print("Hoàn thành tất cả tác vụ")
```

- Bài 6.3: 

```python
import threading

def print_even():
    for i in range(30, 51):
        if i % 2 == 0:
            print("Số chẵn:", i)

def print_odd():
    for i in range(30, 51):
        if i % 2 != 0:
            print("Số lẻ:", i)

t1 = threading.Thread(target=print_even)
t2 = threading.Thread(target=print_odd)

t1.start()
t2.start()

t1.join()
t2.join()
```

- Bài 6.4: 

```python
import threading

def factorial(n):
    result = 1
    for i in range(1, n + 1):
        result *= i
    print(f"Giai thừa của {n} là {result}")

numbers = [5, 6, 7]

threads = []
for n in numbers:
    t = threading.Thread(target=factorial, args=(n,))
    threads.append(t)
    t.start()

for t in threads:
    t.join()
```

- bài 6.7:

```python
import threading
import time
from datetime import datetime

def web_log(name):
    print(f"Starting {name}")
    for _ in range(3):
        print(f"{name}: Web {datetime.now()}")
        time.sleep(1)
    print(f"Exiting {name}")

threads = []
for site in ["Google", "Yahoo", "Facebook"]:
    t = threading.Thread(target=web_log, args=(site,))
    threads.append(t)
    t.start()

for t in threads:
    t.join()

print("Exiting Main Thread")
```

- Bìa 6.8: 

```python
import threading
import random

numbers = [random.randint(0, 10) for _ in range(20)]
total = 0
lock = threading.Lock()

def sum_part(part):
    global total
    s = sum(part)
    lock.acquire()
    total += s
    lock.release()

mid = len(numbers) // 2
t1 = threading.Thread(target=sum_part, args=(numbers[:mid],))
t2 = threading.Thread(target=sum_part, args=(numbers[mid:],))

t1.start()
t2.start()

t1.join()
t2.join()

print("Danh sách:", numbers)
print("Tổng:", total)
```

```python
import threading
import random

numbers = [random.randint(0, 100) for _ in range(30)]
max_value = -1
lock = threading.Lock()

def find_max(part):
    global max_value
    local_max = max(part)
    lock.acquire()
    if local_max > max_value:
        max_value = local_max
    lock.release()

mid = len(numbers) // 2
t1 = threading.Thread(target=find_max, args=(numbers[:mid],))
t2 = threading.Thread(target=find_max, args=(numbers[mid:],))

t1.start()
t2.start()

t1.join()
t2.join()

print("Danh sách:", numbers)
print("Giá trị lớn nhất:", max_value)
```
