```python
# 1: Tạo numpy gtr 0 -> 9, hiện thị phần tử , kiểu dữ liệu, kích thước: 
import numpy as np
import numpy.ma as ma
arr_1 = np.arange(10)

print(f'Các phần tử có trong arr_1: {arr_1}')
print(f'Kiểu sữ liệu arr_1: {arr_1.dtype}')
print(f'Kích thước của arr_1: {arr_1.shape}','\n')

# 2: Tạo arr_odd, arr_even: 
arr_odd = arr_1[arr_1 % 2 != 0]
arr_even = arr_1[arr_1 % 2 == 0]

print(f'Mảng con của arr_1 chứa các phần tử lẻ: {arr_odd}','\n')
print(f'Mảng con của arr_1 chứa các phần tử chẵn: {arr_even}','\n')

# 3: Tạo arr_update với các phần tử chẵn giữ nguyên các phần tử lẻ thay bằng 100
# Cách 1: np.where()
arr_update_1 = np.where(arr_1 % 2 != 0, 100, arr_1)
print(f'Mảng arr_update_1: {arr_update_1}')


# Cách 2: 
arr_update_2 = arr_1
for i in arr_update_2:
    if i % 2 != 0:
        arr_update_2[i] = 100
print(f'Mảng arr_update_2: {arr_update_2}')






```

```python
import numpy as np
arr_a = [1,2,3,4,5,6]

arr_b = [7,2,10,2,7,4,9,4,9,8]
```

```python
# 1 Tạo arr_c chỉ lấy duy nhất các phần tử xuất hiện ở cả arr_a, arr_b :

arr_c = np.intersect1d(arr_a, arr_b) # Phép giao 
print(f'Arr_c chứa các ptu chỉ xuất hiện cả ở arr_a, arr_b: {arr_c}')
# Phép hợp dùng union1d()

# 2: Tạo arr_d chứa các phần tử chỉ xuất hiện ở arr_a: 
arr_d = []
for i in arr_a + arr_b: 
    if i in arr_a: 
        arr_d.append(i)
print(f'Arr_d chỉ chứa các phần tử xuất hiện ở arr_a: {arr_d}')

# 3: 
arr_e = np.array([2,6,1,9,10,3,27,8,6,25,16])
arr_f = []
for i in range(len(arr_e)): 
    if 5 <= arr_e[i] <= 10:
        arr_f.append(arr_e[i])

print(f'Arr_f chỉ chứa các phtu có gtri 5 -> 10: {arr_f}')
```

```python

# Lấy file txt về heights, weights: 
list_heights = []
list_weights = []

# Lấy file heights_1.txt: 
with open('heights_1.txt', 'r') as f: 
    for line in f: 
        list_heights.append(float(line))
print(list_heights)


# Lấy weights_1.txt: 
with open('weights_1.txt', 'r') as f: 
    for line in f: 
        list_weights.append(float(line))
print(list_weights)
```

```python
# 1 : Tạo numpy.array từ heights: 
arr_heights = np.array(list_heights)
print('Danh sách chiều cao: ')
print(arr_heights,'\n')

# 2: Tạo Numpy.array từ weights: 
arr_weights = np.array(list_weights)
print('Danh sách cân nặng: ')
print(arr_weights,'\n')
```

```python
# 3: Quy đổi hệ số chiều cao: 
m = 0.0254
arr_heights_m = arr_heights * m
print('Đổi hệ số từ inch sang đơn vị (mét): ')
print(arr_heights_m)



```

```python
# 4: Quy đổi hệ số cân nặng: 
kg = 0.453592
arr_weights_kg = arr_weights * kg
print('Đổi hệ số từ pound sang đơn vị (kg): ')
print(arr_weights_kg)
```

```python
# 5: Tính BMI: 
arr_bmi = arr_weights_kg/ (arr_heights_m * arr_heights_m)
print('Tính chỉ số BMI: ')
print(arr_bmi)
```

```python
# 6: Gtr ở index = 50 của arr_weights_kg: 
idx = arr_weights_kg[-1]

print(f'Gtri cân nặng ở vtr index = 50: {idx}')
```

```python
# 7: Tạo arr_heights_m_20 gồm các phần tử có idx từ 20 -> 40: 
arr_heights_20 = arr_heights_m[20:41]
print('Mảng gồm các ptu từ index 20 -> 40: ')
print(arr_heights_20)
```

```python
# 8: Cho biết cầu thủ bóng chày có BMI < 21 trong arr_bmi : 
arr_bmi_21 = arr_bmi[arr_bmi < 21]
print('Các cầu thủ bóng chày có bmi < 21: ')
print(arr_bmi_21)
```

```python
# 9: 
# Trung bình chiều cao các cầu thủ: 
heights_tb = sum(arr_heights_m) / len(arr_heights_m)
print(f'Chiều cao TB các cầu thủ: {heights_tb:.2f}')


# Trung bình cân nặng của các cầu thủ: 
weights_tb = sum(arr_weights_kg)/ len(arr_weights_kg)
print((f'Cân nặng TB các cầu thủ : {weights_tb:.2f}'))

```

```python
# 10: Tính chiều cao, cân nặng max : 
max_idx_m = np.argmax(arr_heights_m) # Trả về index chứ không trả về gtr
max_idx_kg = np.argmax(arr_weights_kg)
print(f'Cầu thủ cao nhất: {arr_heights_m[max_idx_m]:.2f}','\n')
print(f'Cầu thủ nặng nhất: {arr_weights_kg[max_idx_kg]:.2f}')

```

```python
# 11: Tính chiều cao, cân nặng min: 
mix_idx_m = np.argmin(arr_heights_m)
min_idx_kg = np.argmin(arr_weights_kg)
print(f'Cầu thủ thấp nhất: {arr_heights_m[mix_idx_m]:.2f}')
print(f'Cầu thủ nhẹ nhất: {arr_weights_kg[min_idx_kg]:.2f}')
```
