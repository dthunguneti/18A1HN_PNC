```python
import pandas as pd
import numpy as np

truth = pd.Series(range(10))

pred = pd.Series(range(10) + np.random.random(10))

```

```python
truth
```

```python
pred
```

- sai số bình phương trung bình là : sai lệch giữa gtri dự đoán và gtri thực 
- Kí hiệu là : MSE

```python
mse = ((truth - pred)**2).mean()

print(f'Sai số bình phương trung bình: {mse}')
```
