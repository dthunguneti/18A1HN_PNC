```python
import pandas as pd
import numpy as np

ser = pd.Series(np.take(list('abcdefgh'), np.random.randint(8, size=30)))
ser
```

- 1 : giải thích đoạn code: 


- list('abcdefgh') : tạo một ds từ kí tứ truyền vào : ['a', 'b', 'c', 'd', 'e', 'f', 'g', 'h']

- np.take() : Lấy phàn tử trong ds tạo từ list() theo chỉ số ngẫu nhiên

- np.random.randint(8, size=30) : tạo 1 mảng có có 30 phần tử ngẫu nhiên từ 0 -> 8 
- pd.Series() : chuyển dữ liêu thành Series Pandas  

```python
ser.value_counts() # Đếm tần suất xuất hiện của ký tự 
# Với index : là gtri (a -> h)
#  giá trị là : số lần suất hiện của các index
```
