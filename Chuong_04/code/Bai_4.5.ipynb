```python
import pandas as pd

import numpy as np

ser = pd.Series(np.random.normal(10, 5, 25)) # Tạo dữ liệu theo phân phối chuẩn Gausian(loc: kỳ vọng, scale: độ lệch chuẩn)
ser
```

- 1 : Tìm giá trị tối thiểu của ser

```python
print(f'Giá trị tối thiểu của ser là: {ser.min():.2f}')
```

- 2: Tìm giá trị tứ phân vị thứ nhất Q1 : 25%

```python
# Cách 1: 
print('Tứ phân vị thứ nhất Q1 giá trị 25%: ',ser.quantile(0.25))

# Cách 2: 

print('Tứ phân vị thứ nhất Q1 giá trị 25%: ',ser.describe()['25%'])
```

- 3: Tìm trung vị của ser : 


```python
# Cách 1: 
print('Tứ phân vị Q2 giá trị 50%: ',ser.quantile(0.5))

# Cách 2: 

print('Tứ phân vị Q2 giá trị 50%: ',ser.describe()['50%'])
```

4: Tìm tứ phân vị Q3 với giá trị 75%: 


```python
# Cách 1: 
print('Tứ phân vị Q3 giá trị 75%: ',ser.quantile(0.75))

# Cách 2: 

print('Tứ phân vị Q3 giá trị 75%: ',ser.describe()['75%'])
```

- 5: giá trị tối đa của ser

```python
print(f'Gía trị tối đa của ser: {ser.max()}')
```
