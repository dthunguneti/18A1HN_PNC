Hàm str.contains() dùng để kiểm tra mỗi chuỗi trong Series có chứa một mẫu ký tự hoặc khớp với biểu thức chính quy hay không, và trả về một Series Boolean.

```python
import pandas as pd

emails = pd.Series(['buying books at amazom.com', 'rameses@egypt.com',
                     'matt@t.co', 'narendra@modi.com'])

pattern = r'[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,4}'

df = emails[emails.str.contains(pattern,regex=True)]
print(df)
```
