```python
import pandas as pd

df_euro_12 = pd.read_csv('../data_raw/euro12.csv')
df_euro_12
```

```python
# IN gtr cột goals: 
print('In giá trị cột Goals: ')
df_euro_12.iloc[:,[0,1]]
```

```python
# Có bao nhiêu đội tham gia Euro2012: 
print('Số đội tham gia Euro_2012: ',df_euro_12['Goals'].count())
```

```python
# Tạo 1 dataframe mới có tên là  discipline chỉ chứa 3 cột 'Team', 'Yellow card', 'Red cards'
a = df_euro_12['Team']
b = df_euro_12['Yellow Cards']
c = df_euro_12['Red Cards']

discipline = pd.DataFrame(pd.concat([a, b, c],axis=1))
print(f'\nDataFrame Discipline: \n{discipline}')
```

```python
# Sắp xếp discipline giảm dần theo cột 'Red Cards': 

a = discipline['Red Cards']

b = a.sort_values()[::-1].index

df = pd.DataFrame(discipline, index=b)
print(f'\nSắp xếp discipline giảm dần theo Red Cards: \n{df}')
```

```python
# Sắp xếp discipline giảm dần theo cột 'Yellow Cards': 

c = discipline['Yellow Cards']

d = c.sort_values()[::-1].index

df_1 = pd.DataFrame(discipline, index=d)
print(f'\nSắp xếp discipline giảm dần theo Yellow Cards: \n{df_1}')
```

```python
# Tính trung bình Yellow Cards :
mean_yellow_cards = df_euro_12['Yellow Cards'].mean()


print(f'Trung bình thẻ vàng là : {mean_yellow_cards}/ 1 trận')
```
