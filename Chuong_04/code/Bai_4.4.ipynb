```python
import pandas as pd

ser_1 = pd.Series([1, 2, 3, 4, 5])

ser_2 = pd.Series([4, 5, 6, 7, 8])

```

- 1: Từ ser_1 hãy xóa các mục có mặt trong ser_2

```python
# Cách 1: 

df_1 = ~ser_1.isin(ser_2)
ser_1[df_1]

```

```python
# Cách 2: Xoá theo nhãn : 
index = ser_1[ser_1.isin(ser_2)].index
df_2 = ser_1.drop(index)

df_2
```

- 2: Lấy tất cả các mục của ser_1 và ser_2 nhưng không nằm chung trong cả hai 

```python
df_3 = ~ser_2.isin(ser_1)

mix = pd.concat([ser_1[df_1], ser_2[df_3]])
mix
```
